#### Starter Code for Question 1

```python
# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right
def is_duplicate(root: TreeNode) -> int:
  seen = set()
    
  def traverse(node):
        if not node:
            return False
        if node.val in seen:
            return True   # Found a duplicate
        seen.add(node.val)
        # Recurse left and right
        return traverse(node.left) or traverse(node.right)
    
  return 1 if traverse(root) else 0
```

```python
# validate the code above
root = TreeNode(5)
root.left = TreeNode(3)
root.right = TreeNode(7, TreeNode(3), TreeNode(8))

print(is_duplicate(root))  # ➜ 1 (duplicate value = 3)
```


-   Copy the solution your partner wrote. 


```python
# Mark's answer 
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    write_pos = 0  # Tracks where to place the next non-zero element
    
    # Iterate through the list
    for i in range(len(nums)):
        if nums[i] != 0:
            # Swap current element with element at write_pos
            nums[i], nums[write_pos] = nums[write_pos], nums[i]
            write_pos += 1

    return nums

example_1 = [0, 1, 0, 3, 12]
print(f"Example 1. Given a list: {example_1}\nThe function output is:", move_zeros_to_end(example_1))  

example_2 = [4, 0, 5, 0, 0, 6]
print(f"\nExample 2. Given a list: {example_2}\n The function output is:", move_zeros_to_end(example_2))  
```


-   Explain why their solution works in your own words.


```python
# Here is why Mark's solution works 
#Idea:
#Use a pointer write_pos to keep track of the position where the next non-zero element should go.

#Loop through the list:

#For each element, check if it’s non-zero.

#If it is, swap it with the element at write_pos.

#Then move write_pos one step forward.

#Effect of swapping:

#Each non-zero element is moved to the front in the same order they appear.

#Every time you place a non-zero element, zeros automatically get pushed toward the end.

#After the loop:

#All non-zero elements have been moved to the start, keeping their order intact.

#All zeros are now grouped at the end.
```
